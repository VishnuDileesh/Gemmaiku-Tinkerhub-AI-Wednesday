```python
import string
import syllables
import json
```

```python
with open('../../data/raw/haikus.json', 'r', encoding='utf-8') as f:
    raw_data = json.load(f)

print(f"Total entries in file: {len(raw_data)}")
```

```python
def get_syllable_count_for_line(line):
    clean_line = line.translate(str.maketrans('', '', string.punctuation))

    count = sum(syllables.estimate(word) for word in clean_line.split())

    return count
```

```python
get_syllable_count_for_line("Master of the cloud")
```

```python
is_perfect_count = 0
```

```python
for i in range(len(raw_data)):
    item = raw_data[i]



    gpt_text = next((turn['value'] for turn in item['conversations'] if turn['from'] == 'gpt'), "")

    lines = [line.strip() for line in gpt_text.split('\n') if line.strip()]

    # print(f"Haiku #{i+1}:")


    counts = []

    for line in lines:
        c = get_syllable_count_for_line(line)
        counts.append(c)
        # print(f"  ({c}) {line}")

    print(f"  Pattern: {counts}")


    if counts == [5, 7, 5]:
        is_perfect_count += 1
        # print("  ✅ PERFECT")

    else:
        pass
        # print("  ❌ IMPERFECT")

    print("-" * 30)
```
